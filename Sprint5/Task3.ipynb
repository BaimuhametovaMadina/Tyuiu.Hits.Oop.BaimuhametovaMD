```
public interface IEmployee
{
    void GetInfo();
    void Work();
    void TakeVacation();
}

// Делегаты
public delegate void SalaryChangedHandler(Employee sender, SalaryChangedEventArgs args);
public delegate void WorkCompletedHandler(Employee sender, string taskName);

public class SalaryChangedEventArgs : EventArgs
{
    public int OldSalary { get; }
    public int NewSalary { get; }

    public SalaryChangedEventArgs(int oldSalary, int newSalary)
    {
        OldSalary = oldSalary;
        NewSalary = newSalary;
    }
}

public abstract class Employee : IEmployee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Salary { get; private set; }

    // Доп. атрибуты
    public string Department { get; set; }
    public DateTime HireDate { get; set; }
    public int PerformanceScore { get; set; }
    public bool IsRemoteWorker { get; set; }

    // События
    public event SalaryChangedHandler SalaryChanged;
    public event WorkCompletedHandler WorkCompleted;

    protected Employee(string name, int age, int salary, string department)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Department = department;
        HireDate = DateTime.Now;
        PerformanceScore = 5;
        IsRemoteWorker = false;
    }

    // Явная реализация интерфейса
    void IEmployee.GetInfo()
    {
        Console.WriteLine($"Базовая информация: {Name}, {Age} лет, зарплата {Salary}");
    }

    void IEmployee.Work()
    {
        Console.WriteLine($"{Name} выполняет базовую работу.");
    }

    void IEmployee.TakeVacation()
    {
        Console.WriteLine($"{Name} берет стандартный отпуск.");
    }

    // Абстрактные методы
    public abstract void GetInfo();
    public abstract void Work();
    public abstract void TakeVacation();

    // Уведомление о смене зарплаты
    protected virtual void OnSalaryChanged(int oldSalary, int newSalary)
    {
        SalaryChanged?.Invoke(this, new SalaryChangedEventArgs(oldSalary, newSalary));
    }

    protected virtual void OnWorkCompleted(string taskName)
    {
        WorkCompleted?.Invoke(this, taskName);
    }

    public void UpdateSalary(int newSalary)
    {
        int old = Salary;
        Salary = newSalary;
        OnSalaryChanged(old, newSalary);
    }

    public double GetExperienceYears()
    {
        return (DateTime.Now - HireDate).TotalDays / 365.0;
    }
}

public class Manager : Employee, IEmployee
{
    public int SubordinatesCount { get; private set; }

    // Доп. атрибуты
    public decimal BonusPercentage { get; set; }
    public string ManagementStyle { get; set; }
    public List<string> CurrentProjects { get; set; }

    public Manager(string name, int age, int salary, int subCount, string department)
        : base(name, age, salary, department)
    {
        SubordinatesCount = subCount;
        BonusPercentage = 0.15m;
        ManagementStyle = "Agile";
        CurrentProjects = new List<string>();
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Подчинённых: {SubordinatesCount}, Отдел: {Department}");
        Console.WriteLine($"Стиль управления: {ManagementStyle}, Проектов: {CurrentProjects.Count}");
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек.");
        OnWorkCompleted("Управление командой");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeVacation(int days)
    {
        Console.WriteLine($"{Name} уходит в отпуск на {days} дней.");
    }

    // Новые методы
    public void AddProject(string projectName)
    {
        CurrentProjects.Add(projectName);
        Console.WriteLine($"{Name} добавил проект '{projectName}' в список.");
    }

    public void ApproveBudget(int amount)
    {
        Console.WriteLine($"{Name} одобряет бюджет в размере {amount} руб.");
    }
}

public class Developer : Employee, IEmployee
{
    public string Specialization { get; set; }

    // Доп. атрибуты
    public int LinesOfCodePerDay { get; set; }
    public List<string> Technologies { get; set; }
    public string CurrentTask { get; set; }

    public Developer(string name, int age, int salary, string specialization, string department)
        : base(name, age, salary, department)
    {
        Specialization = specialization;
        LinesOfCodePerDay = 100;
        Technologies = new List<string>();
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Специализация: {Specialization}, Отдел: {Department}");
        Console.WriteLine($"Технологии: {string.Join(", ", Technologies)}");
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} пишет код по специализации {Specialization}.");
        OnWorkCompleted(CurrentTask ?? "Не указана");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeVacation(Developer otherDev)
    {
        Console.WriteLine($"{Name} спрашивает специализацию у {otherDev.Name}: {otherDev.Specialization}");
        Console.WriteLine($"{Name} идёт в отпуск.");
    }

    // Новые методы
    public void AddTechnology(string tech)
    {
        Technologies.Add(tech);
    }

    public void SetTask(string task)
    {
        CurrentTask = task;
        Console.WriteLine($"{Name} получил задачу: {task}");
    }
}

public class SeniorDeveloper : Developer, IEmployee
{
    public int ExperienceYears { get; set; }

    // Доп. атрибуты
    public bool IsTechLead { get; set; }
    public Dictionary<string, int> MentoredDevelopers { get; set; }

    public SeniorDeveloper(string name, int age, int salary, string specialization, int experienceYears, string department)
        : base(name, age, salary, specialization, department)
    {
        ExperienceYears = experienceYears;
        IsTechLead = false;
        MentoredDevelopers = new Dictionary<string, int>();
        LinesOfCodePerDay = 80;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Опыт работы: {ExperienceYears} лет, TechLead: {IsTechLead}");
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} разрабатывает сложные проекты с опытом {ExperienceYears} лет.");
        OnWorkCompleted("Архитектурная задача");
    }

    public new void TakeVacation()
    {
        Console.WriteLine($"{Name} (Senior Developer) уходит в отпуск с дополнительными бонусами.");
    }

    // Новые методы
    public void MentorDeveloper(string devName, int score)
    {
        MentoredDevelopers[devName] = score;
        Console.WriteLine($"{Name} менторует {devName}, оценка: {score}");
    }
}

public interface ILeader
{
    void LeadTeam();
}

public interface IDesigner
{
    void Design();
}

public class LeadDesigner : Employee, ILeader, IDesigner
{
    public string DesignStyle { get; set; }
    public int TeamSize { get; set; }

    // Доп. атрибуты
    public int DesignsCreated { get; set; }
    public HashSet<string> DesignTools { get; set; }

    public LeadDesigner(string name, int age, int salary, string designStyle, int teamSize, string department)
        : base(name, age, salary, department)
    {
        DesignStyle = designStyle;
        TeamSize = teamSize;
        DesignsCreated = 0;
        DesignTools = new HashSet<string> { "Figma", "Photoshop" };
    }

    public override void GetInfo()
    {
        Console.WriteLine(
            $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Стиль дизайна: {DesignStyle}, Размер команды: {TeamSize}, Отдел: {Department}");
        Console.WriteLine($"Инструменты: {string.Join(", ", DesignTools)}");
    }

    public override void Work()
    {
        LeadTeam();
        Design();
        OnWorkCompleted("Новый макет");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void LeadTeam()
    {
        Console.WriteLine($"{Name} руководит командой из {TeamSize} человек.");
    }

    public void Design()
    {
        Console.WriteLine($"{Name} создает дизайн в стиле {DesignStyle}.");
    }

    // Новый метод
    public void CreateDesign()
    {
        DesignsCreated++;
        Console.WriteLine($"{Name} создал дизайн №{DesignsCreated}");
    }
}

public class EmployeeService
{
    private readonly IEmployee _employee;

    public EmployeeService(IEmployee employee)
    {
        _employee = employee;
    }

    public void ManageEmployee()
    {
        _employee.GetInfo();
        _employee.Work();
        _employee.TakeVacation();
        Console.WriteLine();
    }
}

public class EmployeeManager<T> where T : Employee, IEmployee
{
    private readonly List<T> employees = new List<T>();

    public void AddEmployee(T employee)
    {
        employees.Add(employee);
    }

    public void ShowAllInfo()
    {
        foreach (var emp in employees)
        {
            emp.GetInfo();
            Console.WriteLine();
        }
    }

    public void MakeAllWork()
    {
        foreach (var emp in employees)
        {
            emp.Work();
            Console.WriteLine();
        }
    }
}


static void OnSalaryChanged(Employee sender, SalaryChangedEventArgs args)
{
    Console.WriteLine($"[EVENT] Зарплата {sender.Name} изменилась: {args.OldSalary} -> {args.NewSalary}");
}

static void OnWorkCompleted(Employee sender, string task)
{
    Console.WriteLine($"[EVENT] {sender.Name} завершил задачу: {task}");
}

Manager mgr = new Manager("Анна", 40, 70000, 10, "Управление");
mgr.SalaryChanged += OnSalaryChanged;
mgr.WorkCompleted += OnWorkCompleted;

SeniorDeveloper seniorDev = new SeniorDeveloper("Иван", 30, 80000, "Backend", 8, "Разработка");
seniorDev.SalaryChanged += OnSalaryChanged;
seniorDev.WorkCompleted += OnWorkCompleted;

LeadDesigner leadDesigner = new LeadDesigner("Ольга", 35, 90000, "Минимализм", 5, "Дизайн");
leadDesigner.WorkCompleted += OnWorkCompleted;

Developer dev1 = new Developer("Дмитрий", 28, 60000, "Frontend", "Разработка");
Developer dev2 = new Developer("Елена", 26, 62000, "Fullstack", "Разработка");

mgr.AddProject("ERP System");
mgr.ApproveBudget(500000);
mgr.UpdateSalary(75000);

dev1.AddTechnology("C#");
dev1.AddTechnology(".NET");
dev1.SetTask("Реализация UI");

seniorDev.MentorDeveloper("Дмитрий", 9);
seniorDev.SetTask("Ревью архитектуры");

leadDesigner.CreateDesign();

List<Employee> staff = new List<Employee> { mgr, seniorDev, leadDesigner, dev1, dev2 };

Console.WriteLine("=== Все сотрудники ===");
foreach (var emp in staff)
{
    emp.GetInfo();
    emp.Work();
    emp.TakeVacation();
    Console.WriteLine();
}

EmployeeManager<Employee> employeeManager = new EmployeeManager<Employee>();
employeeManager.AddEmployee(mgr);
employeeManager.AddEmployee(seniorDev);
employeeManager.AddEmployee(dev1);
employeeManager.AddEmployee(leadDesigner);

Console.WriteLine("=== Generic класс - информация ===");
employeeManager.ShowAllInfo();

Console.WriteLine("=== Generic класс - работа ===");
employeeManager.MakeAllWork();

Console.WriteLine("=== Dependency Injection ===");
EmployeeService managerService = new EmployeeService(mgr);
managerService.ManageEmployee();

EmployeeService seniorService = new EmployeeService(seniorDev);
seniorService.ManageEmployee();
```
