```
abstract class Employee {
    public string Name {get; set; }
    public int Age {get; set; }
    public int Salary {get; set; }

    public Employee(string name, int age, int salary) {
        Name = name;
        Age = age;
        Salary = salary;
    }

    abstract public void GetInfo();
    abstract public void Work();
    abstract public void TakeVacation();
}

class Manager : Employee {
    public int SubordinatesCount {get; private set; }
    
    public Manager(string name, int age, int salary) : base(name, age, salary) {
        SubordinatesCount++;
    }
    public override void GetInfo() {
        Console.WriteLine("Имя: " + Name);
        Console.WriteLine("Возраст: " + Age);
        Console.WriteLine("Зарплата: " + Salary);
        Console.WriteLine("Количество подчиненных: " + SubordinatesCount);
    }
    public override void Work() {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек.");
    }
    public override void TakeVacation() {}
}

class Developer : Employee {
    string Specialization;

    public Developer(string name, int age, int salary, string specialization) : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public override void GetInfo() {
        Console.WriteLine("Имя: " + Name);
        Console.WriteLine("Возраст: " + Age);
        Console.WriteLine("Зарплата: " + Salary);
        Console.WriteLine("Специализация: " + Specialization);
    }
    public override void Work() {}
    public override void TakeVacation() {}
    public void TakeVacation(Developer otherDev)
    {
        Console.WriteLine($"{Name} спрашивает специализацию у {otherDev.Name}: {otherDev.Specialization}");
        Console.WriteLine($"{Name} идёт в отпуск.");
    }
}

class Designer : Employee {
    string DesignStyle;

    public Designer(string name, int age, int salary, string designStyle) : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public override void GetInfo() {
        Console.WriteLine("Имя: " + Name);
        Console.WriteLine("Возраст: " + Age);
        Console.WriteLine("Зарплата: " + Salary);
        Console.WriteLine("Стиль: " + DesignStyle);
    }
    public override void Work() {}
    public override void TakeVacation() {}
}

Developer designer1 = new Developer("1", 45, 23444, "Цифровой");
Developer designer2 = new Developer("2", 45, 23444, "Барокко");
designer1.TakeVacation(designer2);
```
