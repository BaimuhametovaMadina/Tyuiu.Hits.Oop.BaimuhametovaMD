```
class Product {
    private string name;
    public string Name {
        get {
            return name;
        }
        set {
            name = value;
        }
    }

    private decimal price;
    public decimal Price {
        get {
            return Price;
        }
        set {
            if (value >= 0) {
                price = value;
            }
        }
    }

    private int quantity;
    public int Quantity {
        get {
            return quantity;
        }
        private set {
            if (value == 0) {
                quantity = 0;
            }
            else {
                quantity += value;
            }
        }
    }

    protected string category;
    public string Category {
        get {
            return category;
        }
        set {
            category = value;
        }
    }
    internal string description;
    public string Description {
        get {
            return description;
        }
        set {
            description = value;
        }
    }

    public Product(string name, decimal price, int quantity, string category, string description) {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;

    }

    public decimal GetTotalPrice() {
        return Price * Quantity;
    }
    public void UpdateQuantity(int amount) {
        Quantity = amount;
    }
    public string GetProduction() {
        return $"Название: {Name}, цена: {Price}, остаток: {Quantity}, категория: {Category}, описание: {Description}";
    }
}

class Seller {
    private string name;
    public string Name {
        get {
            return name;
        }
        set {
            name = value;
        }
    }

    private string employeeId;
    public string EmployeeId {
        get {
            return employeeId;
        }
        private set { employeeId = value; }
    }

    protected decimal salary;
    public decimal Salary {
        get {
            return salary;
        }
        set {
            if (value > 0) {
                salary = value;
            }
        }
    }

    internal string contactInfo;
    public string ContactInfo {
        get {
            return contactInfo;
        }
        set {
            contactInfo = value;
        }
    }

    private List<Product> products = new List<Product>();
    public List<Product> ProductsList() {
        return products;
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo) {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product) {
        products.Add(product);
    }
    public void SellProduct(Product product, int quantity) {
        if (product.Quantity >= quantity && quantity > 0) { 
            product.UpdateQuantity(-quantity); 
            Console.WriteLine($"Продано {quantity} единиц товара {product.Name}. Остаток на складе: {product.Quantity}");
            }
        else if (quantity <= 0) {
            Console.WriteLine("Количество проданных товаров должно быть больше нуля!");
        }
        else {
            Console.WriteLine("Недостаточно товара");
        }
    }
    public string GetSellerInfo() {
        return $"Имя: {Name}, номер сотрудника: {EmployeeId}, зарплата: {Salary}, контактные данные: {ContactInfo}";
    }
}

class Store {
    private string storeName;
    public string StoreName {
        get {
            return storeName;
        }
        set {
            storeName = value;
        }
    }

    private string location;
    public string Location {
        get {
            return location;
        }
        set {
            location = value;
        }
    }

    private string storeHours;
    public string StoreHours {
        get {
            return storeHours;
        }
        set {
            storeHours = value;
        }
    }
    private List<Seller> employees = new List<Seller>();

    public Store(string storeName, string location, string storeHours) {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller) {
        employees.Add(seller);
    }
    public void ListProducts() {
        foreach (Seller employee in employees) {
            foreach (Product product in employee.ProductsList()) Console.WriteLine(product.GetProduction());
        }
    }
    public void GetStoreInfo() {
        Console.WriteLine($"Магазин {StoreName} по адресу {Location}, работает {StoreHours}");
    }
}

class Designer : Employee {
    string DesignStyle;
    public override void GetInfo() {
        Console.WriteLine("Имя: " + Name);
        Console.WriteLine("Возраст: " + Age);
        Console.WriteLine("Зарплата: " + Salary);
        Console.WriteLine("Стиль: " + DesignStyle);
    }
    public override void Work() {}
    public override void TakeVacation() {}
}
```
