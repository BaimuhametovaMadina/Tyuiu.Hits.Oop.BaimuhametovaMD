```
using System;
using System.Collections.Generic;

abstract class Employee {
    public string Name { get; set; }
    public int Age { get; set; }
    public int Salary { get; set; }

    public Employee(string name, int age, int salary) {
        Name = name;
        Age = age;
        Salary = salary;
    }

    public abstract void GetInfo();
    public abstract void Work();
    public abstract void TakeVacation();
}

class Manager : Employee {
    public int SubordinatesCount { get; private set; }

    public Manager(string name, int age, int salary, int subCount) : base(name, age, salary) {
        SubordinatesCount = subCount;
    }

    public override void GetInfo() {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Подчиненных: {SubordinatesCount}");
    }

    public override void Work() {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек.");
    }

    public override void TakeVacation() {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeVacation(int days) {
        Console.WriteLine($"{Name} уходит в отпуск на {days} дней.");
    }
}

class Developer : Employee {
    public string Specialization { get; set; }

    public Developer(string name, int age, int salary, string specialization)
        : base(name, age, salary) {
        Specialization = specialization;
    }

    public override void GetInfo() {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Специализация: {Specialization}");
    }

    public override void Work() {
        Console.WriteLine($"{Name} пишет код по специализации {Specialization}.");
    }

    public override void TakeVacation() {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeVacation(Developer otherDev) {
        Console.WriteLine($"{Name} спрашивает специализацию у {otherDev.Name}: {otherDev.Specialization}");
        Console.WriteLine($"{Name} идёт в отпуск.");
    }
}

class SeniorDeveloper : Developer {
    public int ExperienceYears { get; set; }

    public SeniorDeveloper(string name, int age, int salary, string specialization, int experienceYears)
        : base(name, age, salary, specialization) {
        ExperienceYears = experienceYears;
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Опыт работы: {ExperienceYears} лет");
    }

    public override void Work() {
        Console.WriteLine($"{Name} разрабатывает сложные проекты с опытом {ExperienceYears} лет.");
    }

    public new void TakeVacation() {
        Console.WriteLine($"{Name} (Senior Developer) уходит в отпуск с дополнительными бонусами.");
    }
}

interface ILeader {
    void LeadTeam();
}

interface IDesigner {
    void Design();
}

class LeadDesigner : Employee, ILeader, IDesigner {
    public string DesignStyle { get; set; }
    public int TeamSize { get; set; }

    public LeadDesigner(string name, int age, int salary, string designStyle, int teamSize)
        : base(name, age, salary) {
        DesignStyle = designStyle;
        TeamSize = teamSize;
    }

    public override void GetInfo() {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}, Стиль дизайна: {DesignStyle}, Размер команды: {TeamSize}");
    }

    public override void Work() {
        LeadTeam();
        Design();
    }

    public override void TakeVacation() {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void LeadTeam() {
        Console.WriteLine($"{Name} руководит командой из {TeamSize} человек.");
    }

    public void Design() {
        Console.WriteLine($"{Name} создает дизайн в стиле {DesignStyle}.");
    }
}

class EmployeeManager<T> where T : Employee {
    private List<T> employees = new List<T>();

    public void AddEmployee(T employee) {
        employees.Add(employee);
    }

    public void ShowAllInfo() {
        foreach (var emp in employees) {
            emp.GetInfo();
            Console.WriteLine();
        }
    }

    public void MakeAllWork() {
        foreach (var emp in employees) {
            emp.Work();
            Console.WriteLine();
        }
    }
}

Manager mgr = new Manager("Анна", 40, 70000, 10);
mgr.GetInfo();
mgr.Work();
mgr.TakeVacation();
mgr.TakeVacation(14);

SeniorDeveloper seniorDev = new SeniorDeveloper("Иван", 30, 80000, "Backend", 8);
seniorDev.GetInfo();
seniorDev.Work();
seniorDev.TakeVacation();

LeadDesigner leadDesigner = new LeadDesigner("Ольга", 35, 90000, "Минимализм", 5);
leadDesigner.GetInfo();
leadDesigner.Work();
leadDesigner.TakeVacation();

Developer dev1 = new Developer("Дмитрий", 28, 60000, "Frontend");
Developer dev2 = new Developer("Елена", 26, 62000, "Fullstack");
dev1.TakeVacation(dev2);

List<Employee> staff = new List<Employee> { mgr, seniorDev, leadDesigner, dev1, dev2 };

foreach (var emp in staff) {
    emp.GetInfo();
    emp.Work();
    emp.TakeVacation();
    Console.WriteLine();
}

EmployeeManager<Employee> employeeManager = new EmployeeManager<Employee>();
employeeManager.AddEmployee(mgr);
employeeManager.AddEmployee(seniorDev);
employeeManager.AddEmployee(dev1);
employeeManager.AddEmployee(leadDesigner);

Console.WriteLine("Generic класс - вывод информации о сотрудниках:");
employeeManager.ShowAllInfo();

Console.WriteLine("Generic класс - вызов работы всех сотрудников:");
employeeManager.MakeAllWork();
```
