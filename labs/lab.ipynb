```polyglot-notebook
public interface IReportable {
public string GenerateReport();
}

public interface ITeamLeader {
List<string> GetTeamMembers();
}

class Employee {
    public string Name { get; set; }
    public string Position { get; set; }
    public int Salary { get; set; }

    public Employee(string name, string position, int salary) {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo() {
        Console.WriteLine($"Имя: {Name}, Должность: {Position}, Зарплата: {Salary}");
    }
}

class Manager : Employee {
    public string Department { get; private set; }

    public Manager(string name, string position, int salary, string department) : base(name, position, salary) {
        Department = department;
    }
    public override void PrintInfo() {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

class ProjectManager : Manager, IReportable, ITeamLeader {
    public string ProjectName { get; private set; }
    private List<string> TeamMembers {get; set; }

    public ProjectManager(string name, string position, int salary, string department, string projectName) : base(name, position, salary, department) {
        ProjectName = projectName;
        TeamMembers = new List<string>();
    }
    public override void PrintInfo() {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
    }
    public string GenerateReport() {
        int teamSize = TeamMembers.Count() > 0 ? TeamMembers.Count() : 0;
        return $"Отчет по проекту под названием {ProjectName}, ответственное лицо - {Name}, в команде {teamSize} человек";
    }
    public List<string> GetTeamMembers() {
        return TeamMembers;
    }
    public void AddTeamMember(string memberName) {
        TeamMembers.Add(memberName);
    }
}

Employee worker = new Employee("Вероника", "Уборщица", 20000);
Manager manager = new Manager("Вячеслав", "Менеджер", 50000, "Отдел созвонов");
ProjectManager projectmanager = new ProjectManager("Виктория", "Проджект-менеджер", 40000, "Отдел дизайна", "Калькулятор");

worker.PrintInfo();
manager.PrintInfo();
projectmanager.PrintInfo();

Console.WriteLine(projectmanager.GenerateReport());
projectmanager.AddTeamMember("Кирилл");
projectmanager.AddTeamMember("Елизавета");
List<string> projectMembers = projectmanager.GetTeamMembers();
Console.WriteLine($"Состав команды проджект-менеджера {projectmanager.Name}:");
for (int i = 0; i < projectMembers.Count(); i++) {
    if (i < projectMembers.Count - 1) Console.Write(projectMembers[i] + ", ");
    else Console.WriteLine(projectMembers[i]);
}
Console.WriteLine(projectmanager.GenerateReport());
```
