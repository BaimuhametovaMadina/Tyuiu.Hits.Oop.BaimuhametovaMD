```polyglot-notebook
public interface IInstallable {
    public void Install();
}

public abstract class Component {
    public string Label {get;  set; }
    public string Model {get;  set; }
    public int Price {get;  set; }

    public virtual void GetInfo() {
        Console.WriteLine($"Марка: {Label}, Модель: {Model}, Цена: {Price}");
    }
    public override string ToString() {
        return $"Марка: {Label}, Модель: {Model}, Цена: {Price}";
    }
}

public class CPU : Component, IInstallable {
    public int CoreNumber {get; set; }
    public double Freq {get;  set; }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Кол-во ядер: {CoreNumber}, Частота {Freq} ГГц");
    }

    public void Install() {
        Console.WriteLine($"Процессор {Model} установлен");
    }

    public static bool operator ==(CPU c1, CPU c2) {
        return c1.CoreNumber == c2.CoreNumber && c1.Freq == c2.Freq ? true : false;
    }
    public static bool operator !=(CPU c1, CPU c2) {
        return c1.CoreNumber != c2.CoreNumber || c1.Freq != c2.Freq ? true : false;
    }
}

public class RAM : Component, IInstallable {
    public int Size {get;  set; }
    public double Freq {get;  set; }
    public bool IsDimm {get;  set; }

    public override void GetInfo() {
        base.GetInfo();
        string isdimm = IsDimm ? "DIMM (подходит для установки)" : "отличается от DIMM, не подходит для установки";
        Console.WriteLine($"Объем плашки {Size} ГБ, Частота {Freq} ГГц, форм-фактор {isdimm}");
    }

    public void Install() {
        if (IsDimm) Console.WriteLine($"Плашка памяти размером {Size} ГБ установлена");
        else Console.WriteLine("Форм-фактор плашки должен быть DIMM");
    }
    public static bool operator ==(RAM c1, RAM c2) {
        return c1.Size == c2.Size && c1.Freq == c2.Freq && c1.IsDimm == c2.IsDimm ? true : false;
    }
    public static bool operator !=(RAM c1, RAM c2) {
        return c1.Size != c2.Size || c1.Freq != c2.Freq || c1.IsDimm == c2.IsDimm ? true : false;
    }
}
public class GPU : Component, IInstallable {
    public int MemSize {get;  set; }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Объем памяти: {MemSize} ГБ");
    }

    public void Install() {
        Console.WriteLine($"Видеокарта {Model} установлена");
    }
    public static bool operator ==(GPU c1, GPU c2) {
        return c1.MemSize == c2.MemSize && c1.Model == c2.Model ? true : false;
    }
    public static bool operator !=(GPU c1, GPU c2) {
        return c1.MemSize != c2.MemSize || c1.Model != c2.Model ? true : false;
    }
}

public class HDD : Component, IInstallable {
    public int Size {get;  set; }
    public double WriteSpeed {get;  set; }
    public double ReadSpeed {get;  set; }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Объем диска: {Size} ГБ, скорость записи {WriteSpeed} Мб/с, скорость чтения {ReadSpeed} Мб/с");
    }

    public void Install() {
        Console.WriteLine($"Жесткий диск объемом {Size} Гб установлен");
    }
    public static bool operator ==(HDD c1, HDD c2) {
        return c1.Size == c2.Size && c1.WriteSpeed == c2.WriteSpeed && c1.ReadSpeed == c2.ReadSpeed ? true : false;
    }
    public static bool operator !=(HDD c1, HDD c2) {
        return c1.Size != c2.Size || c1.WriteSpeed != c2.WriteSpeed || c1.ReadSpeed != c2.ReadSpeed ? true : false;
    }
}

public class Computer
{
    private readonly List<Component> components = new();
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        Console.WriteLine($"Добавлен компонент: {component.Label} {component.Model}");
    }
    
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
            Console.WriteLine($"Удален компонент: {component.Label} {component.Model}");
        else
            Console.WriteLine($"Компонент не найден: {component.Label} {component.Model}");
    }
    
    public decimal GetTotalPrice() => components.Sum(c => c.Price);
    public void ShowComponents()
    {
        Console.WriteLine("Компоненты компьютера:");
        foreach (var comp in components)
            Console.WriteLine(comp);
    }

    public void InstallAll<T>() where T : Component, IInstallable
    {
    Console.WriteLine($"Установка всех компонентов типа {typeof(T).Name}...");
    foreach (var c in components.OfType<T>())
    {
        c.Install();
    }
    }
    
    public void InstallAll()
    {
        Console.WriteLine("Установка всех компонентов...");
        foreach (var c in components)
        {
            if (c is IInstallable installable)
                installable.Install();
        }
    }
}

public static class ComputerFactory
{
    
    public static Computer CreateGamingComputer()
    {
        var comp = new Computer();
        comp.AddComponent(new CPU { Label = "Intel", Model = "Core i9", Price = 55000, CoreNumber = 12, Freq = 2.7 });
        comp.AddComponent(new RAM { Label = "Corsair", Model = "Corsair 16GB", Price = 8000, Size = 16, Freq = 1666, IsDimm = true });
        comp.AddComponent(new GPU { Label = "NVIDIA", Model = "RTX 3080", Price = 70000, MemSize = 10 });
        comp.AddComponent(new HDD { Label = "WD", Model ="Blue 1TB", Size = 1000, Price = 5000, WriteSpeed = 45000, ReadSpeed = 46600 });
        return comp;
    }
}

var computer = ComputerFactory.CreateGamingComputer();

computer.ShowComponents();

Console.WriteLine($"Общая стоимость: {computer.GetTotalPrice()} руб.");

var ramToRemove = new RAM { Label = "Corsair", Model = "Corsair 16GB", Price = 8000, Size = 16, Freq = 1666, IsDimm = true };
computer.RemoveComponent(ramToRemove);

computer.ShowComponents();
Console.WriteLine($"Общая стоимость после удаления: {computer.GetTotalPrice()} руб.");

computer.InstallAll();
```
